- 
Mean pooling compresses each sentence into a single static vector, which removes word order and local contextual information. As a result, the MLP can only rely on averaged semantic content and cannot effectively capture negation, modifiers, or word interactions, limiting its expressive power.
- What advantage did the LSTM’s sequential processing provide over the MLP?

The LSTM models sentences as sequences, allowing it to capture word order, contextual dependencies, and long-range interactions. This enables the model to better handle negation, sentiment shifts, and temporal dependencies, leading to improved performance over the MLP on sentiment classification tasks.
